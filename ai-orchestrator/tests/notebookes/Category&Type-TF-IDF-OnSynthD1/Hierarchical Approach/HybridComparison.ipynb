```python
import joblib
from pathlib import Path

model_dir = Path("../models")

# Type classifier trained on 500 rows
type_model = joblib.load(model_dir / "type_500.joblib")

# Category classifiers trained on 5200 rows
expense_model = joblib.load(model_dir / "expense_5200.joblib")
income_model = joblib.load(model_dir / "income_5200.joblib")

def predict_hybrid(text):
    predicted_type = type_model.predict([text])[0]

    if predicted_type == "Expense":
        predicted_category = expense_model.predict([text])[0]
    else:
        predicted_category = income_model.predict([text])[0]

    return {
        "type": predicted_type,
        "category": predicted_category
    }

print(predict_hybrid(
    "I earned 4500 EGP designing a website"
))
```

```python
import joblib
from pathlib import Path

model_dir = Path("../models")

models = {
    "500 + 500": {
        "type": joblib.load(model_dir / "type_500.joblib"),
        "expense": joblib.load(model_dir / "expense_500.joblib"),
        "income": joblib.load(model_dir / "income_500.joblib")
    },
    "5200 + 5200": {
        "type": joblib.load(model_dir / "type_5200.joblib"),
        "expense": joblib.load(model_dir / "expense_5200.joblib"),
        "income": joblib.load(model_dir / "income_5200.joblib")
    },
    "500 + 5200 (Hybrid)": {
        "type": joblib.load(model_dir / "type_500.joblib"),
        "expense": joblib.load(model_dir / "expense_5200.joblib"),
        "income": joblib.load(model_dir / "income_5200.joblib")
    },
    "5200 + 500 (Reverse)": {
        "type": joblib.load(model_dir / "type_5200.joblib"),
        "expense": joblib.load(model_dir / "expense_500.joblib"),
        "income": joblib.load(model_dir / "income_500.joblib")
    }
}

def predict_with_models(text, configuration):
    m = models[configuration]
    predicted_type = m["type"].predict([text])[0]

    category_model = (
        m["expense"] if predicted_type == "Expense"
        else m["income"]
    )

    predicted_category = category_model.predict([text])[0]

    return predicted_type, predicted_category
```

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score
)

comparison = []
all_predictions = {}

for name in models:

    predictions = [
        predict_with_models(text, name)
        for text in test_df["description"]
    ]

    predicted_types = [p[0] for p in predictions]
    predicted_categories = [p[1] for p in predictions]

    all_predictions[name] = predictions

    type_acc = accuracy_score(
        test_df["true_type"], predicted_types
    )

    category_acc = accuracy_score(
        test_df["true_category"], predicted_categories
    )

    category_f1 = f1_score(
        test_df["true_category"],
        predicted_categories,
        average="macro",
        zero_division=0
    )

    joint_acc = sum(
        t == true_t and c == true_c
        for (t, c), true_t, true_c in zip(
            predictions,
            test_df["true_type"],
            test_df["true_category"]
        )
    ) / len(test_df)

    comparison.append({
        "Configuration": name,
        "Type Accuracy": type_acc,
        "Category Accuracy": category_acc,
        "Category Macro-F1": category_f1,
        "Joint Accuracy": joint_acc
    })

comparison_df = pd.DataFrame(comparison)

display(
    comparison_df.style.format({
        "Type Accuracy": "{:.2%}",
        "Category Accuracy": "{:.2%}",
        "Category Macro-F1": "{:.2%}",
        "Joint Accuracy": "{:.2%}"
    })
)
```

```python
hybrid_name = "500 + 5200 (Hybrid)"

hybrid_predictions = all_predictions[hybrid_name]

errors = test_df.copy()

errors["predicted_type"] = [
    p[0] for p in hybrid_predictions
]

errors["predicted_category"] = [
    p[1] for p in hybrid_predictions
]

errors = errors[
    (errors["true_type"] != errors["predicted_type"]) |
    (errors["true_category"] != errors["predicted_category"])
]

display(errors)
```

```python
from sklearn.metrics import accuracy_score, f1_score
import pandas as pd

final_comparison = []
final_predictions = {}

for name in models:
    predictions = [
        predict_with_models(text, name)
        for text in final_test["description"]
    ]

    pred_types = [p[0] for p in predictions]
    pred_categories = [p[1] for p in predictions]

    final_predictions[name] = predictions

    type_acc = accuracy_score(final_test["type"], pred_types)
    type_f1 = f1_score(
        final_test["type"], pred_types,
        average="macro", zero_division=0
    )

    category_acc = accuracy_score(
        final_test["category"], pred_categories
    )
    category_f1 = f1_score(
        final_test["category"], pred_categories,
        average="macro", zero_division=0
    )

    joint_acc = sum(
        t == true_t and c == true_c
        for (t, c), true_t, true_c in zip(
            predictions,
            final_test["type"],
            final_test["category"]
        )
    ) / len(final_test)

    final_comparison.append({
        "Configuration": name,
        "Type Accuracy": type_acc,
        "Type Macro-F1": type_f1,
        "Category Accuracy": category_acc,
        "Category Macro-F1": category_f1,
        "Joint Accuracy": joint_acc
    })

final_results = pd.DataFrame(final_comparison)

display(final_results.style.format({
    col: "{:.2%}"
    for col in final_results.columns
    if col != "Configuration"
}))
```
